## Problem 1: Representing SHA-256 Data

### 32-bit Words

SHA-256 uses 32-bit words during its calculations.

In Python, a 32-bit word can be represented using NumPy's `uint32` type.

A 32-bit unsigned integer can store values from:

$$
0 \leq x \leq 2^{32} - 1
$$

For this example, one of the initial SHA-256 values, `0x6A09E667`, will be stored as a `numpy.uint32`.

The value will then be displayed in decimal, hexadecimal and binary form.

### References

[1] NIST, *FIPS 180-4: Secure Hash Standard*  
https://nvlpubs.nist.gov/nistpubs/FIPS/NIST.FIPS.180-4.pdf

[2] McLoughlin, I., *Computational Theory - Binary Representations*  
https://github.com/ianmcloughlin/computational-theory/blob/main/notes/binary-representations.ipynb

```python
import numpy as np

word = np.uint32(0x6A09E667)

print("Decimal:", word)
print("Hexadecimal:", hex(word))
print("Binary:", f"{word:032b}")
```

### Sequences of 32-bit Words

SHA-256 also works with sequences of 32-bit words.

A sequence can contain multiple 32-bit values that are processed together during the algorithm.

In Python, a sequence of 32-bit words can be represented using a NumPy array with the data type `uint32`.

This makes sure that each value in the array is stored as an unsigned 32-bit integer.

For example, some of the initial SHA-256 values can be stored together in one sequence.

#### References

[1] NIST, *FIPS 180-4: Secure Hash Standard*.  
https://nvlpubs.nist.gov/nistpubs/FIPS/NIST.FIPS.180-4.pdf

[2] McLoughlin, I., *Computational Theory - Binary Representations*.  
https://github.com/ianmcloughlin/computational-theory/blob/main/notes/binary-representations.ipynb

```python
import numpy as np

# Sequence of 32-bit words
words = np.array([
    0x6A09E667,
    0xBB67AE85,
    0x3C6EF372,
    0xA54FF53A
], dtype=np.uint32)

print(words)

for word in words:
    print("Hexadecimal:", hex(int(word)))
```

### Input Messages

SHA-256 can process messages of different lengths.

In Python, an input message can be represented using the `bytes` type. This allows the message to be handled as a sequence of byte values before it is processed by SHA-256.

A normal Python string can be converted into bytes using `.encode()`.

For example, the message `"abc"` can be converted into bytes before being used by the algorithm.

#### References

[1] NIST, *FIPS 180-4: Secure Hash Standard*.  
https://nvlpubs.nist.gov/nistpubs/FIPS/NIST.FIPS.180-4.pdf

[2] McLoughlin, I., *Computational Theory - Hash Functions*.  
https://github.com/ianmcloughlin/computational-theory/blob/main/notes/hash-functions.ipynb

```python
# Example input message
message = "abc"

# Convert the message into bytes
message_bytes = message.encode()

print("Message:", message)
print("Bytes:", message_bytes)
print("Length:", len(message_bytes))
```

### 512-bit Message Blocks

SHA-256 processes messages in blocks of 512 bits.

Since one byte contains 8 bits, a 512-bit block contains:

$$
512 \div 8 = 64
$$

bytes.

In Python, a 512-bit message block can therefore be represented using a `bytes` object containing 64 bytes.

Later in SHA-256, each 512-bit block can also be divided into sixteen 32-bit words.

For this example, a block containing 64 zero bytes will be created.

#### References

[1] NIST, *FIPS 180-4: Secure Hash Standard*.  
https://nvlpubs.nist.gov/nistpubs/FIPS/NIST.FIPS.180-4.pdf

[2] McLoughlin, I., *Computational Theory - Hash Functions*.  
https://github.com/ianmcloughlin/computational-theory/blob/main/notes/hash-functions.ipynb

```python
# Create a 512-bit message block
block = bytes(64)

print("Block:", block)
print("Bytes:", len(block))
print("Bits:", len(block) * 8)
```

### Final 256-bit Hash Value

SHA-256 produces a final hash value that is 256 bits long.

Since one byte contains 8 bits, the final hash contains:

$$
256 \div 8 = 32
$$

bytes.

A SHA-256 hash is commonly displayed in hexadecimal form. Since each hexadecimal digit represents 4 bits, a 256-bit hash contains 64 hexadecimal digits.

In Python, the final hash value can be represented using a `bytes` object and displayed using hexadecimal.

For this example, the SHA-256 hash value for `"abc"` will be used.

#### References

[1] NIST, *FIPS 180-4: Secure Hash Standard*.  
https://nvlpubs.nist.gov/nistpubs/FIPS/NIST.FIPS.180-4.pdf

[2] McLoughlin, I., *Computational Theory - Hash Functions*.  
https://github.com/ianmcloughlin/computational-theory/blob/main/notes/hash-functions.ipynb

```python
# Example final SHA-256 hash value
hash_value = bytes.fromhex(
    "ba7816bf8f01cfea414140de5dae2223"
    "b00361a396177a9cb410ff61f20015ad"
)

print("Hash:", hash_value.hex())
print("Bytes:", len(hash_value))
print("Bits:", len(hash_value) * 8)
```

### Big-Endian Integers

SHA-256 uses big-endian byte ordering when working with integer values.

Big-endian means that the most significant byte is stored first.

For example, the 32-bit value:

`0x12345678`

can be split into four bytes:

`12 34 56 78`

In Python, an integer can be converted into bytes using big-endian byte order.

The bytes can also be converted back into an integer.

#### References

[1] NIST, *FIPS 180-4: Secure Hash Standard*.  
https://nvlpubs.nist.gov/nistpubs/FIPS/NIST.FIPS.180-4.pdf

[2] McLoughlin, I., *Computational Theory - Binary Representations*.  
https://github.com/ianmcloughlin/computational-theory/blob/main/notes/binary-representations.ipynb

```python
# Example 32-bit integer
value = 0x12345678

# Convert the integer into 4 bytes using big-endian order
big_endian = value.to_bytes(4, byteorder="big")

print("Integer:", hex(value))
print("Big-endian bytes:", big_endian.hex())

# Convert the bytes back into an integer
converted_value = int.from_bytes(big_endian, byteorder="big")

print("Converted back:", hex(converted_value))
```
