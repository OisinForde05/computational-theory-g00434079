### 32-bit Words

SHA-256 performs most of its calculations using 32-bit words.

A 32-bit unsigned word can represent values in the range:

$$
0 \leq x \leq 2^{32} - 1
$$

In Python, NumPy's `uint32` type can be used to represent these values.

Using `uint32` is suitable because SHA-256 requires fixed-width unsigned 32-bit words. A normal Python integer can grow beyond 32 bits, whereas `uint32` keeps the value represented using 32 bits.

Hexadecimal is also useful when displaying SHA-256 words because one hexadecimal digit represents 4 bits. Therefore, a 32-bit word can be displayed using 8 hexadecimal digits.

For example, one of the initial SHA-256 values is:

`0x6A09E667`

SHA-256 uses 32-bit words and arithmetic modulo $2^{32}$
([FIPS 180-4, Sections 3.1 and 3.2](https://nvlpubs.nist.gov/nistpubs/FIPS/NIST.FIPS.180-4.pdf#page=12)).

Binary and hexadecimal representations are also discussed in
[McLoughlin's Binary Representations notebook](https://github.com/ianmcloughlin/computational-theory/blob/main/notes/binary-representations.ipynb).

```python
import numpy as np

word = np.uint32(0x6A09E667)

print("Decimal:", word)
print("Hexadecimal:", hex(word))
print("Binary:", f"{word:032b}")
```

This ensures that each value in the array is represented as an unsigned 32-bit integer.

For example, several of the initial SHA-256 values can be stored together in one array.

A SHA-256 message block is represented as sixteen 32-bit words

bytes.

A SHA-256 hash is commonly displayed in hexadecimal form.

Since one hexadecimal digit represents 4 bits, a 256-bit hash contains 64 hexadecimal digits.

In Python, the final hash value can be represented using a `bytes` object and displayed in hexadecimal form.

### Big-Endian Integers

SHA-256 uses big-endian byte ordering when interpreting integer values.

Big-endian means that the most significant byte is stored first.

For example, the 32-bit value:

`0x12345678`

can be separated into four bytes:

`12 34 56 78`

In Python, an integer can be converted into bytes using big-endian byte order. The bytes can also be converted back into the original integer.

Specifying the byte order is important because the same sequence of bytes could represent a different integer if it was interpreted using a different byte order.

Using big-endian therefore ensures that Python interprets the values in the same order required by SHA-256.

The big-endian representation used by SHA-256 is described in
[FIPS 180-4, Section 3.1](https://nvlpubs.nist.gov/nistpubs/FIPS/NIST.FIPS.180-4.pdf#page=12).

## Problem 2: SHA-256 Bitwise Operations

### Ch(x, y, z)

The `Ch` function is one of the Boolean functions used by SHA-256.

It chooses bits from either `y` or `z` depending on the value of the corresponding bit in `x`.

The function is defined as:

$$
Ch(x, y, z) = (x \land y) \oplus (\neg x \land z)
$$

In Python, the bitwise operators `&`, `~` and `^` can be used to implement this operation.

NumPy `uint32` values are used so that the inputs are represented as unsigned 32-bit integers.

The SHA-256 definition of `Ch` is given in
[FIPS 180-4, Section 4.1.2](https://nvlpubs.nist.gov/nistpubs/FIPS/NIST.FIPS.180-4.pdf#page=15).

The use of binary values and bitwise operations in Python is also demonstrated in
[McLoughlin's Binary Representations notebook](https://github.com/ianmcloughlin/computational-theory/blob/main/notes/binary-representations.ipynb).

```python
import numpy as np


def Ch(x, y, z):
    """Return the SHA-256 choice function."""
    return np.uint32((x & y) ^ (~x & z))
```

### Maj(x, y, z)

The `Maj` function is another Boolean function used by SHA-256.

It looks at the corresponding bits of `x`, `y` and `z` and returns the value that appears in the majority.

The function is defined as:

$$
Maj(x, y, z) = (x \land y) \oplus (x \land z) \oplus (y \land z)
$$

In Python, the bitwise operators `&` and `^` can be used to implement this operation.

NumPy `uint32` values are used so that the inputs are represented as unsigned 32-bit integers.

The SHA-256 definition of `Maj` is given in
[FIPS 180-4, Section 4.1.2](https://nvlpubs.nist.gov/nistpubs/FIPS/NIST.FIPS.180-4.pdf#page=15).

The use of binary values and bitwise operations in Python is also demonstrated in
[McLoughlin's Binary Representations notebook](https://github.com/ianmcloughlin/computational-theory/blob/main/notes/binary-representations.ipynb).

```python
def Maj(x, y, z):
    """Return the SHA-256 majority function."""
    return np.uint32((x & y) ^ (x & z) ^ (y & z))
```

### Sigma0(x)

The `Sigma0` function is one of the bitwise functions used by SHA-256.

It performs three right rotations on the input value and combines the results using XOR.

The function is defined as:

$$
\Sigma_0^{256}(x) =
ROTR^2(x) \oplus ROTR^{13}(x) \oplus ROTR^{22}(x)
$$

A right rotation moves bits to the right while moving any bits that leave the end back to the beginning.

In Python, the rotations are built using the bitwise shift operators `>>` and `<<`, while `|` combines the shifted parts and `^` combines the three rotation results.

NumPy `uint32` values are used so that the value remains represented as an unsigned 32-bit integer.

The SHA-256 definition of `Sigma0` is given in
[FIPS 180-4, Section 4.1.2](https://nvlpubs.nist.gov/nistpubs/FIPS/NIST.FIPS.180-4.pdf#page=15).

The use of binary values and bitwise operations in Python is also demonstrated in
[McLoughlin's Binary Representations notebook](https://github.com/ianmcloughlin/computational-theory/blob/main/notes/binary-representations.ipynb).

```python
def Sigma0(x):
    """Return the SHA-256 Sigma0 function."""

    rotate_2 = (x >> 2) | (x << 30)
    rotate_13 = (x >> 13) | (x << 19)
    rotate_22 = (x >> 22) | (x << 10)

    return np.uint32(rotate_2 ^ rotate_13 ^ rotate_22)
```

### Sigma1(x)

The `Sigma1` function is another bitwise function used by SHA-256.

It performs three right rotations on the input value and combines the results using XOR.

The function is defined as:

$$
\Sigma_1^{256}(x) =
ROTR^6(x) \oplus ROTR^{11}(x) \oplus ROTR^{25}(x)
$$

A right rotation moves bits to the right while moving any bits that leave the end back to the beginning.

In Python, the rotations are built using the bitwise shift operators `>>` and `<<`. The operator `|` combines the shifted parts, while `^` combines the three rotation results.

NumPy `uint32` values are used so that the value remains represented as an unsigned 32-bit integer.

The SHA-256 definition of `Sigma1` is given in
[FIPS 180-4, Section 4.1.2](https://nvlpubs.nist.gov/nistpubs/FIPS/NIST.FIPS.180-4.pdf#page=15).

The use of binary values and bitwise operations in Python is also demonstrated in
[McLoughlin's Binary Representations notebook](https://github.com/ianmcloughlin/computational-theory/blob/main/notes/binary-representations.ipynb).

```python
def Sigma1(x):
    """Return the SHA-256 Sigma1 function."""

    rotate_6 = (x >> 6) | (x << 26)
    rotate_11 = (x >> 11) | (x << 21)
    rotate_25 = (x >> 25) | (x << 7)

    return np.uint32(rotate_6 ^ rotate_11 ^ rotate_25)
```

### sigma0(x)

The `sigma0` function is one of the bitwise functions used by SHA-256.

It performs two right rotations and one right shift on the input value. The three results are then combined using XOR.

The function is defined as:

$$
\sigma_0^{256}(x) =
ROTR^7(x) \oplus ROTR^{18}(x) \oplus SHR^3(x)
$$

A right rotation moves bits to the right while moving any bits that leave the end back to the beginning.

A right shift also moves bits to the right, but the bits that leave the end are discarded instead of being moved back to the beginning.

In Python, the rotations are created using `>>`, `<<` and `|`, while the final right shift only requires `>>`. The `^` operator combines the three results.

NumPy `uint32` values are used so that the value remains represented as an unsigned 32-bit integer.

The SHA-256 definition of `sigma0` is given in
[FIPS 180-4, Section 4.1.2](https://nvlpubs.nist.gov/nistpubs/FIPS/NIST.FIPS.180-4.pdf#page=15).

The use of binary values and bitwise operations in Python is also demonstrated in
[McLoughlin's Binary Representations notebook](https://github.com/ianmcloughlin/computational-theory/blob/main/notes/binary-representations.ipynb).

```python
def sigma0(x):
    """Return the SHA-256 sigma0 function."""

    rotate_7 = (x >> 7) | (x << 25)
    rotate_18 = (x >> 18) | (x << 14)
    shift_3 = x >> 3

    return np.uint32(rotate_7 ^ rotate_18 ^ shift_3)
```

### sigma1(x)

The `sigma1` function is one of the bitwise functions used by SHA-256.

It performs two right rotations and one right shift on the input value. The three results are then combined using XOR.

The function is defined as:

$$
\sigma_1^{256}(x) =
ROTR^{17}(x) \oplus ROTR^{19}(x) \oplus SHR^{10}(x)
$$

A right rotation moves bits to the right while moving any bits that leave the end back to the beginning.

A right shift moves bits to the right, but the bits that leave the end are discarded.

In Python, the rotations are created using `>>`, `<<` and `|`, while the right shift only requires `>>`. The `^` operator combines the three results.

NumPy `uint32` values are used so that the value remains represented as an unsigned 32-bit integer.

The SHA-256 definition of `sigma1` is given in
[FIPS 180-4, Section 4.1.2](https://nvlpubs.nist.gov/nistpubs/FIPS/NIST.FIPS.180-4.pdf#page=15).

The use of binary values and bitwise operations in Python is also demonstrated in

```python
def sigma1(x):
    """Return the SHA-256 sigma1 function."""

    rotate_17 = (x >> 17) | (x << 15)
    rotate_19 = (x >> 19) | (x << 13)
    shift_10 = x >> 10

    return np.uint32(rotate_17 ^ rotate_19 ^ shift_10)
```

## Problem 3: Generating the SHA-256 Constants

### Generating Prime Numbers

The SHA-256 round constants are based on the first 64 prime numbers.

A prime number is a whole number greater than 1 that is only divisible by 1 and itself.

For this part, a function called `primes(n)` will generate the first `n` prime numbers.

The function checks each number starting from 2. If the number has no divisors other than 1 and itself, it is added to the list of prime numbers.

SHA-256 uses the first 64 prime numbers when generating its round constants
([FIPS 180-4, Section 4.2.2](https://nvlpubs.nist.gov/nistpubs/FIPS/NIST.FIPS.180-4.pdf)).

Prime numbers and methods for working with them in Python are also discussed in
[McLoughlin's Prime Numbers notebook](https://github.com/ianmcloughlin/computational-theory/blob/main/notes/prime-numbers.ipynb).

```python
def primes(n):
    """Return the first n prime numbers."""

    prime_numbers = []
    number = 2

    while len(prime_numbers) < n:
        is_prime = True

        for divisor in range(2, number):
            if number % divisor == 0:
                is_prime = False
                break

        if is_prime:
            prime_numbers.append(number)

        number += 1

    return prime_numbers
```

### Calculating Cube Roots of the First 64 Primes

The SHA-256 round constants are generated from the cube roots of the first 64 prime numbers.

The `primes(64)` function from the previous step is used to generate the prime numbers.

NumPy's `cbrt()` function can then calculate the cube root of each value.

For example:

$$
\sqrt[3]{2} \approx 1.259921
$$

The fractional part of each cube root will be used in the next step.

The SHA-256 round constants are based on the cube roots of the first 64 prime numbers
([FIPS 180-4, Section 4.2.2](https://nvlpubs.nist.gov/nistpubs/FIPS/NIST.FIPS.180-4.pdf)).

Working with roots is also discussed in
[McLoughlin's Roots notebook](https://github.com/ianmcloughlin/computational-theory/blob/main/notes/roots.ipynb).

```python
# Generate the first 64 prime numbers
prime_numbers = np.array(primes(64))

# Calculate the cube root of each prime
cube_roots = np.cbrt(prime_numbers)

print("First 10 cube roots:")
print(cube_roots[:10])
```

### Extracting the Fractional Parts

The next step is to extract the fractional part of each cube root.

For example:

$$
\sqrt[3]{2} \approx 1.259921
$$

The whole-number part is `1`, leaving the fractional part:

$$
1.259921 - 1 = 0.259921
$$

In Python, `numpy.floor()` can be used to find the whole-number part of each cube root. This value can then be subtracted from the original cube root.

The SHA-256 round constants are generated from the fractional parts of the cube roots of the first 64 prime numbers
([FIPS 180-4, Section 4.2.2](https://nvlpubs.nist.gov/nistpubs/FIPS/NIST.FIPS.180-4.pdf)).

The `floor()` operation used to remove the whole-number part is described in the
[NumPy `floor` documentation](https://numpy.org/devdocs/reference/generated/numpy.floor.html).

Working with roots is also discussed in
[McLoughlin's Roots notebook](https://github.com/ianmcloughlin/computational-theory/blob/main/notes/roots.ipynb).

```python
# Extract the fractional part of each cube root
fractional_parts = cube_roots - np.floor(cube_roots)

print("First 10 fractional parts:")
print(fractional_parts[:10])
```

### Converting to 32-bit Unsigned Integers

The fractional parts calculated in the previous step must now be converted into 32-bit unsigned integers.

Each fractional value is multiplied by:

$$
2^{32}
$$

The remaining fractional part is then removed, leaving an integer that can be stored as a 32-bit unsigned value.

For example, the fractional part of the cube root of 2 is approximately:

$$
0.259921
$$

This value is multiplied by $2^{32}$ and the integer part is kept.

NumPy's `uint32` type is then used to store the resulting value as an unsigned 32-bit integer.

The SHA-256 round constants are generated using this process
([FIPS 180-4, Section 4.2.2](https://nvlpubs.nist.gov/nistpubs/FIPS/NIST.FIPS.180-4.pdf)).

The removal of the remaining fractional part is performed using
[NumPy `floor`](https://numpy.org/devdocs/reference/generated/numpy.floor.html).

```python
# Multiply the fractional parts by 2^32
scaled_values = fractional_parts * (2**32)

# Remove the decimal part and convert to uint32
round_constants = np.floor(scaled_values).astype(np.uint32)

print("First 10 values:")
print(round_constants[:10])
```

## Problem 4: Padding and Parsing Messages

### Basic Message Padding

Before SHA-256 can process a message, the message must be padded.

The original length of the message is first recorded in bits.

A `1` bit is then appended to the message. Since the message is represented using bytes, this can be added using `0x80`, which has the binary representation:

`10000000`

Zero bytes are then added until the padded message reaches 56 bytes within the final 64-byte block.

The remaining 8 bytes will later be used to store the original message length.

SHA-256 message padding is defined in
[FIPS 180-4, Section 5.1.1](https://nvlpubs.nist.gov/nistpubs/FIPS/NIST.FIPS.180-4.pdf#page=18).

Message padding is also discussed in
[McLoughlin's Padding notebook](https://github.com/ianmcloughlin/computational-theory/blob/main/notes/padding.ipynb).

```python
def block_parse(msg):
    """Begin padding a SHA-256 message."""

    # Record the original message length in bits
    message_length = len(msg) * 8

    # Append the 1 bit
    padded = msg + b"\x80"

    # Add zero bytes until 8 bytes remain in the final block
    while len(padded) % 64 != 56:
        padded += b"\x00"

    return padded
```

```python
message = b"abc"

padded = block_parse(message)

print("Original message:", message)
print("Original length in bits:", len(message) * 8)
print("Padded length so far:", len(padded))
```
