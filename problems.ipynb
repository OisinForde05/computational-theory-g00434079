## Problem 1: Representing SHA-256 Data

### 32-bit Words

SHA-256 uses 32-bit words during its calculations.

In Python, a 32-bit word can be represented using NumPy's `uint32` type.

A 32-bit unsigned integer can store values from:

$$
0 \leq x \leq 2^{32} - 1
$$

For this example, one of the initial SHA-256 values, `0x6A09E667`, will be stored as a `numpy.uint32`.

The value will then be displayed in decimal, hexadecimal and binary form.

### References

[1] NIST, *FIPS 180-4: Secure Hash Standard*  
https://nvlpubs.nist.gov/nistpubs/FIPS/NIST.FIPS.180-4.pdf

[2] McLoughlin, I., *Computational Theory - Binary Representations*  
https://github.com/ianmcloughlin/computational-theory/blob/main/notes/binary-representations.ipynb

```python
import numpy as np

word = np.uint32(0x6A09E667)

print("Decimal:", word)
print("Hexadecimal:", hex(word))
print("Binary:", f"{word:032b}")
```
